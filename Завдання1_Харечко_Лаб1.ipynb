### Порівняння типів даних:
У зчитаному наборі даних переважають такі типи:
* **int64** (числові цілі значення): для колонок `id`, `score`, `money`.
* **object** (текстові / строкові значення): для колонок `name`, `category`.

Використання методів `info()` дозволяє швидко визначити наявність пропущених значень та обсяг пам'яті, яку займає датафрейм.

```python
import pandas as pd
import numpy as np

data = {
    'id': list(range(1, 11)),
    'name': ['Item_'+str(i) for i in range(1, 11)],
    'quantity': np.random.randint(10, 100, size=10),
    'money': np.random.randint(1000, 5000, size=10)
}
df = pd.DataFrame(data)
df.to_csv('test_data.csv', index=False)
```

## 1. Базовий огляд датафрейму



```python
print("--- head() ---")
display(df.head())

print("\n--- shape ---")
print(df.shape)

print("\n--- info() ---")
df.info()

print("\n--- describe() ---")
display(df.describe())
```

## 2. Перетворення колонки в масив NumPy та обчислення

```python
money_array = df['money'].to_numpy()

mean_money = money_array.mean()

print("Масив money:", money_array)
print(f"Середнє значення колонки 'money': {mean_money:.2f}")
```
