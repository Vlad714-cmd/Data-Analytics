```python
df = pd.read_csv('messi barca.csv', encoding='latin1')
print(df.head())
```

```python
goals_by_season = df.groupby('Season')['Goals scored'].sum()
print(goals_by_season)
```

```python
avg_minutes = df.groupby('Season')['Minutes played'].mean()
print(avg_minutes)
```

```python
matches_count = df.groupby(['Season', 'Competition']).size()
print(matches_count)
```

```python
matches_count_df = df.groupby(['Season', 'Competition']).size().reset_index(name='count')
print(matches_count_df.head(10))
```

```python
summary = df.groupby('Competition')['Goals scored'].agg(['sum', 'mean', 'min', 'max'])
print(summary)
```

```python
pivot = pd.pivot_table(df, values='Goals scored', index='Competition', columns='Age', aggfunc='sum', fill_value=0)
print(pivot)
```

```python
crosstab = pd.crosstab(df['Season'], df['Competition'])
print(crosstab)
```

```python
importance_df = pd.DataFrame({
    'Competition': df['Competition'].unique(),
    'Importance': ['High' if 'League' in str(c) or 'Champions' in str(c) else 'Medium' for c in df['Competition'].unique()]
})

df_merged = df.merge(importance_df, on='Competition', how='left')
print(df_merged.head())
```

```python
half = len(df) // 2
df_part1 = df.iloc[:half]
df_part2 = df.iloc[half:]

df_concat = pd.concat([df_part1, df_part2])
print(f"Розміри: {len(df_part1)} + {len(df_part2)} = {len(df_concat)}")
```

```python
left = pd.DataFrame({'A': [1, 2, 3]}, index=['a', 'b', 'c'])
right = pd.DataFrame({'B': [10, 20, 30]}, index=['a', 'b', 'c'])

joined = left.join(right)
print(joined)
```

```python
total_goals = df['Goals scored'].sum()
goals_share = (df.groupby('Competition')['Goals scored'].sum() / total_goals * 100).round(2)
print(goals_share)
```

```python
df_teams = pd.DataFrame({
    'team_id': [1, 2],
    'team_name': ['FC Barcelona', 'Paris Saint-Germain'],
    'country': ['Spain', 'France']
})

df_player_matches = pd.DataFrame({
    'match_id': [101, 102, 103, 104],
    'team_id': [1, 1, 2, 2],
    'goals': [2, 1, 0, 3],
    'rating': [8.5, 7.2, 6.0, 9.1]
})
```

```python
df_merged_custom = df_player_matches.merge(df_teams, on='team_id', how='left')
```

```python
aggregated_custom = df_merged_custom.groupby('team_name')['goals'].agg(['sum', 'mean', 'count'])
print("Агреговані показники:\n", aggregated_custom)
```

```python
pivot_custom = pd.pivot_table(df_merged_custom, values='rating', index='team_name', columns='goals', aggfunc='mean', fill_value=0)
print("\nЗведена таблиця:\n", pivot_custom)
```

```python
pivot_custom.to_csv('football_summary.csv')
```
