```python
import pandas as pd

def group_and_sort_by_column(df, column_name):
    result_df = df[column_name].value_counts().reset_index()

    result_df.columns = [column_name, 'count']

    result_df = result_df.sort_values(by='count', ascending=False)

    return result_df

competition_stats = group_and_sort_by_column(df, 'Competition')

print("=== Результат групування та сортування Competition ===")
print(competition_stats)

competition_stats.to_csv('/content/drive/MyDrive/competition_summary.csv', index=False)
print("\nРезультат успішно збережено у файл 'competition_summary.csv'")
```
